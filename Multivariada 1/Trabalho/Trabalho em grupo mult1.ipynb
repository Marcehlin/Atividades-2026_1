### *(d) cálculo e interpretação da matriz de variâncias e covariâncias amostral;*

A matriz de variâncias e covariâncias amostral é:

|Matriz VarCov          | **Pressão Sistólica**  | **Colesterol**  | **Glicemia**    |
|-----------------------|------------------------|-----------------|-----------------|
| **Pressão Sistólica** |     133.399            |     128.877     |      36.293     |

| **Glicemia**          |     36.293             |     75.331      |      80.389     |

Interpretação:

1. As variâncias não são comparáveis, pois as escalas das variáveis são diferentes:

- Glicemia varia 72.27  até 113.67 unidades (amplitude = 41.4 unidades);

2. As covariâncias amostrais são positivas, o que indica uma relação linear positiva entre as variáveis, i.e., as variáveis tendem a variar no mesmo sentido (se uma aumentar outras tendem a aumentar também, e se uma diminuir as outras tendem a diminuir).

### *(e) cálculo e interpretação da matriz de correlações amostral;*

A matriz de correlações amostral é:

|Matriz Corr            | **Pressão Sistólica**  | **Colesterol**  | **Glicemia**    |
|-----------------------|------------------------|-----------------|-----------------|
| **Pressão Sistólica** |     1                  |     0.622       |      0.350      |
| **Colesterol**        |     0.622              |     1           |      0.468      |
| **Glicemia**          |     0.350              |     0.468       |      1          |

Interpretação:

Com base na amostra observada de 65 indivíduos retirada da população desconhecida do contexto, estima-se que a correlação linear 

- entre a Pressão Sistólica e o Colesterol seja de 0.622, sugere uma relação linear postiva moderada entre as variáveis
- entre a Pressão Sistólica e a Glicemia seja de 0.350, sugere uma relação linear postiva fraca entre as variáveis
- entre o Colesterol e a Glicemia seja de 0.468, sugere uma relação linear postiva fraca entre as variáveis

Dessa forma, há indicativos de que, na população estudada:

indivíduos com pressão sistólica maior tendem (moderadamente) a apresentar nível de colesterol mais alto; indivíduos com pressão sistólica maior tendem a apresentar nível de glicemia mais alto; indivíduos com nível de colesterol maior tendem a apresentar nível de glicemia mais alto;
